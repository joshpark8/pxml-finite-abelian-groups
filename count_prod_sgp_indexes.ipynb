# Constructing group

```python
invariants = [2, 2, 2]  # change this to e.g. [4, 6] or [2, 3, 5]
G = AdditiveAbelianGroup(invariants)
N = G.order()


def elt(i):
    """Integer i → element of G, MSB-first mixed-radix.
    For Z_2^n this is exactly the binary expansion of i."""
    coords = []
    for j, n in enumerate(invariants):
        weight = prod(invariants[j + 1 :])
        coords.append((i // weight) % n)
    return G(coords)


# Cyclic subgroup H_i = <elt(i)>
H = {i: G.submodule([elt(i)]) for i in range(N)}
for i in H:
    print(f"H{i}", H[i].gens())


def to_int(g):
    """Convert a group element to its integer index using the mixed-radix coordinates."""
    v = g.lift()  # length matches len(invariants)
    i = 0
    for j, n in enumerate(invariants):
        weight = prod(invariants[j + 1 :])
        i += (int(v[j]) % n) * weight
    return i


def sort_key(H):
    """Sorted tuple of nonzero-element indices."""
    return tuple(sorted(to_int(g) for g in H if g != G.zero()))


# Enumerate all order-4 subgroups
order4 = set()
for i in range(1, N):
    for j in range(i + 1, N):
        K = G.submodule([elt(i), elt(j)])
        if K.order() == 4:
            order4.add(K)

# Sort lexicographically
K = sorted(order4, key=sort_key)

for sgp in range(len(K)):
    print(f"K{sgp+1}", K[sgp].gens())
```
